```python
from voice_ai.tts import generate_speech
from voice_ai.stt import transcribe_audio

sample_text = "Hey! It's me your new voice assistant. I am working perfectly."
print(f"Original Text: {sample_text}")

audio_path = generate_speech(sample_text, "test_voice.mp3")
print(f"Audio saved to: {audio_path}")

print("Listening to the audio file and transcribing...")
transcribed_text = transcribe_audio("test_voice.mp3")

print(f"Transcribed Text: {transcribed_text}")
```
